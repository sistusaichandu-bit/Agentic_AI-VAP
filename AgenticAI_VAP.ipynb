```python
df = pd.read_csv('property_prices_100.csv')
print(df)
```

```python
%matplotlib inline
plt.xlabel('Area')
plt.ylabel('Price')
plt.scatter(df.area, df.price, color='red', marker='*')
```

```python
new_area = df.drop('price', axis='columns')
new_area
```

```python
new_price = df.drop('area', axis='columns')
new_price
```

```python
reg = linear_model.LinearRegression()
reg.fit(new_area, new_price)
```

```python
reg.predict([[3000]])
```

```python
reg.predict([[5000]]) # This approx value is given by regression models
```

```python
df_predict = pd.read_csv('areas.csv')
```
